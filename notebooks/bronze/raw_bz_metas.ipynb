```python
df_metas_br = read(
    base_path=RAW_PATH,
    table_name=METAS_BR
)
```

```python
df_metas_municipio = read(
    base_path=RAW_PATH,
    table_name=METAS_MUNICIPIO
)
```

```python
df_metas_uf = read(
    base_path=RAW_PATH,
    table_name=METAS_UF
)
```

```python
df_meta_silver_br = (
    df_metas_br
    # =============================
    # Conversão de tipos
    # =============================
    .withColumn(
        "ANO",
        col("ano").cast("int")
    )
    .withColumn(
        "TAXA_ALFABETIZACAO",
        col("taxa_alfabetizacao").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2024",
        col("meta_alfabetizacao_2024").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2025",
        col("meta_alfabetizacao_2025").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2026",
        col("meta_alfabetizacao_2026").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2027",
        col("meta_alfabetizacao_2027").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2028",
        col("meta_alfabetizacao_2028").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2029",
        col("meta_alfabetizacao_2029").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2030",
        col("meta_alfabetizacao_2030").cast("double")
    )
    .withColumn(
        "PERCENTUAL_PARTICIPACAO",
        col("percentual_participacao").cast("double")
    )
    # =============================
    # Padronização
    # =============================
    .withColumn(
        "REDE",
        initcap(trim(col("rede")))
    )
    # =============================
    # Chave técnica
    # =============================
    .withColumn(
        "SK_META_ALFABETIZACAO",
        sha2(
            concat_ws(
                "|",
                col("ANO"),
                upper(col("REDE"))
            ),
            256
        )
    )
    # =============================
    # Auditoria
    # =============================
    .withColumn(
        "DT_PROCESSAMENTO",
        current_date()
    )
    .withColumn(
        "TS_PROCESSAMENTO",
        current_timestamp()
    )
)
```

```python
df_meta_silver_br_selected = df_meta_silver_br.select(
    "SK_META_ALFABETIZACAO",
    "ANO",
    "REDE",
    "TAXA_ALFABETIZACAO",
    "META_ALFABETIZACAO_2024",
    "META_ALFABETIZACAO_2025",
    "META_ALFABETIZACAO_2026",
    "META_ALFABETIZACAO_2027",
    "META_ALFABETIZACAO_2028",
    "META_ALFABETIZACAO_2029",
    "META_ALFABETIZACAO_2030",
    "PERCENTUAL_PARTICIPACAO",
    "DT_PROCESSAMENTO",
    "TS_PROCESSAMENTO"
)
```

```python
df_meta_municipio_silver = (
    df_metas_municipio
    # =============================
    # Conversão de tipos
    # =============================
    .withColumn("ANO", col("ano").cast("int"))
    .withColumn(
        "CO_MUNICIPIO",
        col("id_municipio").cast("int")
    )
    .withColumn(
        "TAXA_ALFABETIZACAO",
        col("taxa_alfabetizacao").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2024",
        col("meta_alfabetizacao_2024").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2025",
        col("meta_alfabetizacao_2025").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2026",
        col("meta_alfabetizacao_2026").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2027",
        col("meta_alfabetizacao_2027").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2028",
        col("meta_alfabetizacao_2028").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2029",
        col("meta_alfabetizacao_2029").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2030",
        col("meta_alfabetizacao_2030").cast("double")
    )
    .withColumn(
        "NIVEL_ALFABETIZACAO",
        col("nivel_alfabetizacao").cast("int")
    )
    .withColumn(
        "PERCENTUAL_PARTICIPACAO",
        col("percentual_participacao").cast("double")
    )
    # =============================
    # Padronização
    # =============================
    .withColumn(
        "REDE",
        initcap(trim(col("rede")))
    )
    # =============================
    # Chave Técnica
    # =============================
    .withColumn(
        "SK_META_MUNICIPIO",
        sha2(
            concat_ws(
                "|",
                col("ANO"),
                col("CO_MUNICIPIO"),
                upper(col("REDE"))
            ),
            256
        )
    )

    # =============================
    # Auditoria
    # =============================

    .withColumn(
        "DT_PROCESSAMENTO",
        current_date()
    )

    .withColumn(
        "TS_PROCESSAMENTO",
        current_timestamp()
    )

)
```

```python
df_meta_municipio_silver_selected = (
    df_meta_municipio_silver
    .select(
        "SK_META_MUNICIPIO",
        "ANO",
        "CO_MUNICIPIO",
        "REDE",
        "TAXA_ALFABETIZACAO",
        "META_ALFABETIZACAO_2024",
        "META_ALFABETIZACAO_2025",
        "META_ALFABETIZACAO_2026",
        "META_ALFABETIZACAO_2027",
        "META_ALFABETIZACAO_2028",
        "META_ALFABETIZACAO_2029",
        "META_ALFABETIZACAO_2030",
        "NIVEL_ALFABETIZACAO",
        "PERCENTUAL_PARTICIPACAO",
        "DT_PROCESSAMENTO",
        "TS_PROCESSAMENTO"
    )
)
```

```python
df_meta_estado_silver = (
    df_metas_uf
    # =============================
    # Conversão de tipos
    # =============================

    .withColumn(
        "ANO",
        col("ano").cast("int")
    )
    .withColumn(
        "TAXA_ALFABETIZACAO",
        col("taxa_alfabetizacao").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2024",
        col("meta_alfabetizacao_2024").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2025",
        col("meta_alfabetizacao_2025").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2026",
        col("meta_alfabetizacao_2026").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2027",
        col("meta_alfabetizacao_2027").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2028",
        col("meta_alfabetizacao_2028").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2029",
        col("meta_alfabetizacao_2029").cast("double")
    )
    .withColumn(
        "META_ALFABETIZACAO_2030",
        col("meta_alfabetizacao_2030").cast("double")
    )
    .withColumn(
        "PERCENTUAL_PARTICIPACAO",
        col("percentual_participacao").cast("double")
    )
    # =============================
    # Padronização
    # =============================
    .withColumn(
        "SG_UF",
        upper(trim(col("sigla_uf")))
    )
    .withColumn(
        "REDE",
        initcap(trim(col("rede")))
    )
    # =============================
    # Chave Técnica
    # =============================
    .withColumn(
        "SK_META_ESTADO",
        sha2(
            concat_ws(
                "|",
                col("ANO"),
                col("SG_UF"),
                upper(col("REDE"))
            ),
            256
        )
    )

    # =============================
    # Auditoria
    # =============================

    .withColumn(
        "DT_PROCESSAMENTO",
        current_date()
    )

    .withColumn(
        "TS_PROCESSAMENTO",
        current_timestamp()
    )

)
```

```python
df_meta_estado_silver_selected = (
    df_meta_estado_silver
    .select(
        "SK_META_ESTADO",
        "ANO",
        "SG_UF",
        "REDE",
        "TAXA_ALFABETIZACAO",
        "META_ALFABETIZACAO_2024",
        "META_ALFABETIZACAO_2025",
        "META_ALFABETIZACAO_2026",
        "META_ALFABETIZACAO_2027",
        "META_ALFABETIZACAO_2028",
        "META_ALFABETIZACAO_2029",
        "META_ALFABETIZACAO_2030",
        "PERCENTUAL_PARTICIPACAO",
        "DT_PROCESSAMENTO",
        "TS_PROCESSAMENTO"
    )
)
```

```python
write_delta(
    df=df_meta_silver_br_selected,
    base_path=BRONZE_PATH,
    table_name=METAS_BR,
    merge_keys=["SK_META_ALFABETIZACAO"]
)
```

```python
write_delta(
    df=df_meta_municipio_silver_selected,
    base_path=BRONZE_PATH,
    table_name=METAS_MUNICIPIO,
    merge_keys=["SK_META_MUNICIPIO"]
)
```

```python
write_delta(
    df=df_meta_estado_silver_selected,
    base_path=BRONZE_PATH,
    table_name=METAS_UF,
    merge_keys=["SK_META_ESTADO"]
)
```
